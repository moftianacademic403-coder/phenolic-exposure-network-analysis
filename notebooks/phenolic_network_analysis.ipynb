# Phenolic Exposure–Effect Network Analysis

This notebook runs the complete reproducible workflow for BPA, triclosan (TCS), and triclocarban (TCC) versus 31 biological effect biomarkers. The analysis includes within-chemical FDR correction, network connectivity metrics, 1,000-resample bootstrap stability, covariate-adjusted partial Spearman correlations, and a cross-validated graphical LASSO sensitivity analysis.

## 1. Configure paths
Place the participant-level workbook at `data/total.xlsx`, or change `INPUT_FILE` below. Generated files are written to `outputs/`.

```python
from pathlib import Path
import subprocess
import sys

REPOSITORY_ROOT = Path.cwd().resolve()
if REPOSITORY_ROOT.name == 'notebooks':
    REPOSITORY_ROOT = REPOSITORY_ROOT.parent

INPUT_FILE = REPOSITORY_ROOT / 'data' / 'total.xlsx'
OUTPUT_DIR = REPOSITORY_ROOT / 'outputs'
SCRIPT = REPOSITORY_ROOT / 'src' / 'phenolic_network_analysis.py'

print('Input:', INPUT_FILE)
print('Output:', OUTPUT_DIR)
print('Script:', SCRIPT)
```

## 2. Run the complete analysis
The manuscript settings are fixed below: exposed-group code 1, 1,000 bootstrap resamples, random seed 42, FDR alpha 0.05, and robust-core threshold 0.70.

```python
command = [
    sys.executable,
    str(SCRIPT),
    '--input', str(INPUT_FILE),
    '--sheet', 'total',
    '--output', str(OUTPUT_DIR),
    '--exposed-group', '1',
    '--bootstrap', '1000',
    '--seed', '42',
    '--fdr-alpha', '0.05',
    '--robust-threshold', '0.70',
]

completed = subprocess.run(command, check=True, text=True, capture_output=True)
print(completed.stdout)
```
